2. **What is an interpreter in Python ?**
-> An interpreter in Python is a program that reads and executes Python code line by line.
It converts the high-level Python code into machine code that your computer can understand.
Unlike a compiler, it doesn’t translate the whole program at once—this helps with easier debugging.
The Python interpreter runs scripts directly and is built into environments like the Python shell or IDEs.

5. **What is Mutability in Python?**
-> Mutability refers to the ability of an object to be changed after it has been created. In Python, some data types are mutable, meaning you can change their contents, while others are immutable, meaning once created, their value cannot be altered.

Mutable Examples: list, dict, set, bytearray
Immutable Examples: int, float, str, tuple, bool, frozenset.

6. **Why Lists Are Mutable and Tuples Are Immutable ?**
-> Lists are mutable because they are designed for dynamic, changeable data — you can add, remove, or modify elements.

Tuples are immutable because they are intended for fixed, constant data — once created, they cannot be changed.

Lists use more memory but allow flexibility; tuples are lightweight and safe for use as dictionary keys or in secure data.

This design helps Python balance performance, safety, and functionality.

7. **What is the difference between “==” and “is” operators in Python ?**
-> The difference between == and is in Python comes down to comparison of values versus comparison of object identity.

**== (Equality Operator)**
Compares values of two objects.

Returns True if the values of the objects are the same, even if they are different objects in memory.

**is (Identity Operator)**
Compares memory locations (identities) of two objects.

Returns True if both objects refer to the same object in memory.

8. **What Are Logical Operators in Python?**
-> Logical operators in Python are used to combine conditional statements or Boolean expressions. They help evaluate multiple conditions and return True or False based on the logic of the conditions.

There are three main logical operators in Python:

**and -**
Returns True if both conditions are True.
If either condition is False, it returns False.

**or -**
Returns True if at least one condition is True.
Only returns False if both conditions are False.

**not -**
Reverses the Boolean value of the given condition.
If the condition is True, it becomes False.
If the condition is False, it becomes True.

9. **What is Type Casting in Python?**
-> Type casting refers to the process of converting one data type to another. In Python, you can change a value from one type (such as integer, float, string) to another type. This is important because operations often require data to be in a certain type, and type casting helps ensure compatibility between different data types.

Python allows two types of type casting:

Implicit Type Casting (Automatic Type Conversion)

Explicit Type Casting (Manual Type Conversion)



10. **Difference Between Implicit and Explicit Type Casting in Python ?**
-> **Implicit Type Casting (Automatic Type Conversion)**
**Definition**: The Python interpreter automatically converts one data type to another when required, without the programmer's intervention.

When It Happens: It usually occurs when Python converts from a lower data type to a higher data type (e.g., int to float).

No manual intervention required.

**Explicit Type Casting (Manual Type Conversion)**
**Definition**: The programmer manually converts one data type to another using built-in functions such as int(), float(), str(), etc.

When It Happens: This is done when you need to forcefully convert a value to a specific type, even if Python wouldn't automatically do it.

Manual intervention required.


11. **What Is the Purpose of Conditional Statements in Python?**
-> Conditional statements in Python are used to make decisions in your program based on whether a condition is True or False. They allow your code to take different paths or perform different actions depending on the situation.

**Purpose of Conditional Statements:**
**Control the flow of the program**

They let your program decide which block of code to run.

**Implement logic and decision-making**

For example, "If the user is over 18, allow access."

**Execute code selectively**

Only run certain code when specific conditions are met.

**Improve interactivity and intelligence**

Programs respond differently based on input, values, or user choices.

**Conditional Keywords in Python:-**

if: Checks a condition.

elif: Checks another condition if the previous one is False.

else: Executes if all conditions are False.

12. **How Does the elif Statement Work in Python?**
-> The elif statement in Python stands for "else if" and is used to check multiple conditions after an initial if statement. It allows your program to test several different conditions one after the other, and execute the block of code for the first condition that is True.

**Purpose of elif:**
Used when you have more than two possible paths or choices.

Helps avoid writing multiple nested if statements.

Only one block of code will run, even if multiple conditions are True — Python executes the first matching condition and then exits the block.

13. **Difference Between for and while Loops in Python ?**
-> Both for and while loops are used to repeat blocks of code, but they are used in different scenarios based on the control logic.

**For Loop:-**
Used when the number of iterations is known.

Iterates over sequences like list, tuple, string, or range.

Automatically handles the loop variable and its progression.

Preferred for data structure traversal.

Less risk of infinite loops (handled by sequence end).

Compact and clean syntax.

Commonly used in counting and iterating tasks.

**While Loop :-**
Used when the number of iterations is unknown.

Runs as long as a given condition remains True.

Requires manual management of the loop variable.

Suitable for logic-based repetition (e.g., waiting for user input).

Can easily lead to infinite loops if the condition is not properly handled.

More flexible for dynamic stop conditions.

Commonly used in real-time checks or user-driven loops.

14. **Describe a scenario where a while loop is more suitable than a for loop.**
-> User Login System:-

Imagine you're building a simple login system where a user is allowed to try logging in until they enter the correct password. You don't know how many attempts they'll need — it could be 1 or 100 — so a while loop is ideal.

**while Is Better Here:-**
You don't know in advance how many times the user will try.

The loop runs until a condition is met (user_input == correct_password).

Using a for loop would require guessing the number of attempts, which is inefficient or unsafe.

```python
correct_password = "python123"
user_input = ""

while user_input != correct_password:
    user_input = input("Enter password: ")

print("Access granted.")
```
